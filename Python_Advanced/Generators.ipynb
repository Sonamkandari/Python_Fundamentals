```python
# numbers = range(1,100000000)
# Imagine storing
# 100 million numbers.
# Huge memory.

# -> heres comes the Generator Idea

# -> why store everything
# -> What if I create numbers only when you ask?"

# Technical Definition : A generator is a special type of iterator that generates values lazily, meaning one value at a time, only when requested, instead of creating and storing all values in the memory at once. 
```

```python
# Normal function
def numbers():
    return [1,2,3]

# what happens?

numbers()  # -> returns everything

```

```python
# Normal function
def numbers():
    yield 1
    yield 2
    yield 3
    yield 4

# what happens?
gen = numbers()
print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))

print()
# -> or Alternative is
for num in numbers():
    print (num)

```

```python
# Generate a fibonacci series using a generator

def fibo(n):
    a = 0
    b = 1
    for _ in range (n):
        yield a
        a = b
        b = a+b

for num in  fibo(6):
    print(num)        
    
```

```python
def fibo(n):
    if n<= 0:
        return 0
    elif n == 1:
        return 1

    return fibo(n-1)+fibo(n-2)


for i in range (7):
    print(fibo(i))
```

```python
numbers = [10,20,30] # Iterable
it = iter(numbers) # Iterator

print(next(it))
print(next(it))
print(next(it))  # When there are no more elements, next(it) raises StopIteration.
```
