```python
# we can create our  own class

# It will be divided into two parts 
# the first one will be the creation of the class 
class Item:
    # Methods means functions that are inside the classes
    def calculate_total_price(self,x,y):
        return x*y 

item1 = Item() # this step is equal to creating a instance or a object of the class
# we are allowed to assign some attributes to the objects of the class
item1.name1 = "Fiction Books"
item1.price = 599
item1.quantity = 600

# calling the method
value = item1.calculate_total_price(item1.price,item1.quantity)
print("Total item of first item: ",value)

# creating another instance of the class
item2 = Item() # this step is equal to creating a instance or a object of the class
# we are allowed to assign some attributes to the objects of the class
item2.name1 = "Fiction Books"
item2.name2 = "Non-fiction Books" 
item2.price = 550
item2.quantity = 500

value = item2.calculate_total_price(item2.price,item2.quantity)
print("Total price of second item: ",value)

```

```python
class Item:
    def __init__(self,name,price,quantity):  # constructor
        self.name = name
        self.price = price
        self.quantity= quantity

    def calculate_total_price(self, x, y):
        return x * y

item1 = Item("Non-Fiction books",100,30)
 
item2 = Item("Fictions books",100,20)


print(item1.name)
print(item1.price)
print(item1.quantity)

print(item2.name)
print(item2.price)
print(item2.quantity)
```

```python
from xml.parsers.expat import model


class Car:
    # to construct the car object we need a special type of methods called constructor
    def __init__(self, model, year, color, for_sale): # this is our constructor method, we need this method in order to construct objects
        self.model = model
        self.year = year
        self.color = color
        self.for_sale = for_sale

    def drive(self):
        print(f"sonam drive the {self.model}")    

    def stop(self):
        print(f"you stop driving the car {self.model}")  

    def describe(self):
        print(f"{self.year} {self.color} {self.model}")      
# Car class is already defined in this notebook, so no external import is needed
# to construct a car object we need a unique name
car1 = Car("BMW",2004,"Red",True)
car2 = Car("Mustang",2005,"Yellow",False)

print()
print(car1.model,car1.year, car1.color, car1.for_sale)
print(car2.model,car2.year, car2.color, car2.for_sale)

print()
car1.drive()
car1.stop()

print()
car2.drive()
car2.stop()

print()
car1.describe()
car2.describe()



```
