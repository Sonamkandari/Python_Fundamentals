```python
# Assignment

a = 9
b = a

print(a)
print(b)

# there is one integer object

# Both variable a and b are refer to it
# Because a and b point to the same object.


listA = [1,2,3,4,5]
listB = listA

print(listA)
print(listB)

listB.append(8)
# Because listA and ListB point to the same list object.
print(listA)
```

```python
# Shallow copy
# -> A shallow copy creates a new outer object but copies references to nested objects instead of copying them .

import copy
a = [[1,2],[3,4]]
b = copy.copy(a)

print(b)

# The outer list is new  -> the inner list are shared
a[0].append(100)
print(a)
print(b) # Because both copies share the same inner list.


```

```python
# Deep Copy 
import copy
a = [[1,2], [3,4]]
b = copy.deepcopy(a)

print (b)
print(a[0].append(700))
print(a)
print(b)


# | Feature               | Assignment (`=`) | Shallow Copy | Deep Copy |
# | --------------------- | ---------------- | ------------ | --------- |
# | New outer object      | ❌                | ✅            | ✅         |
# | New nested objects    | ❌                | ❌            | ✅         |
# | Shares nested objects | ✅                | ✅            | ❌         |
# | Memory usage          | Lowest           | Medium       | Highest   |
# | Speed                 | Fastest          | Faster       | Slowest   |
```
