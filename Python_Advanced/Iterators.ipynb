```python
# we cannot fully understand the generators until you understand Iterators

# An Iterator is an object that allows sequential traversal of elements one at a time.
# It implements the iterator protocol by providing the _itr_() -next_()

numbers = [10, 20, 30]   # Iterable

it = iter(numbers)        # Iterator

print(next(it))           # 10
print(next(it))           # 20
print(next(it))           # 30
```
