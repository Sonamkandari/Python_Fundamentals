```python
class Animal:  # this a parent class and there are some methods of parent class
    def __init__(self,name):
        self.name = name
        self.is_alive = True

    def eat(self):
        print(f"{self.name} is eating")

    def sleep(self):
        print(f"{self.name} is sleeping")    


class Dog(Animal):  # class dog inherit from Animal
    def __init__(self,name):
            self.name = name
            self.is_alive = True
    
    def eat(self):
            print(f"{self.name} is eating")
    
    def sleep(self):
            print(f"{self.name} is sleeping")  

class Cat(Animal):
    def __init__(self,name):
            self.name = name
            self.is_alive = True
    
    def eat(self):
            print(f"{self.name} is eating")
    
    def sleep(self):
            print(f"{self.name} is sleeping")     

class Mouse(Animal):
    def __init__(self,name):
            self.name = name
            self.is_alive = True
    
    def eat(self):
            print(f"{self.name} is eating")
    
    def sleep(self):
            print(f"{self.name} is sleeping")  



# so if we will make any change some where in the method then we have to manually make changes every where therefore we use inheritance 
# which makes it easy to write code once and use it multiple time .. enhance code reusability                


dog = Dog("Coco")
cat = Cat("Garfield")
mouse = Mouse("Micky")



print(dog.name)
print(dog.is_alive)
dog.sleep()
dog.eat()

print()
print(cat.name)
print(cat.is_alive)
cat.eat()
cat.sleep()

print()

print(mouse.name)
print(mouse.is_alive)
mouse.eat()
mouse.sleep()
    
```

```python
class Animal:  # this a parent class and there are some methods of parent class
    def __init__(self,name):
        self.name = name
        self.is_alive = True

    def eat(self):
        print(f"{self.name} is eating")

    def sleep(self):
        print(f"{self.name} is sleeping")    


class Dog(Animal):  # class dog inherit from Animal
    pass

class Cat(Animal):
    pass

class Mouse(Animal):
    pass


dog = Dog("Coco")
cat = Cat("Garfield")
mouse = Mouse("Micky")



print(dog.name)
print(dog.is_alive)
dog.sleep()
dog.eat()

print()
print(cat.name)
print(cat.is_alive)
cat.eat()
cat.sleep()

print()

print(mouse.name)
print(mouse.is_alive)
mouse.eat()
mouse.sleep()
    
```

```python
# Multiple Inheritance = a child class inherit the traits from more then one parent class
# for example their are two parent classes, father and mother and one child class 

class Mother:
    def __init__(self,hair_color,height,skin_tone):
        self.hair_color = hair_color
        self.height = height
        self.skin_tone = skin_tone

class Father:
    def __init__(self,eye_color):
        self.eye_color = eye_color


class Child(Mother,Father):
    def __init__(self,hair_color, height, eye_color, name, skin_tone):
       Mother.__init__ (self,hair_color,height,skin_tone)
       Father.__init__(self,eye_color)
       self.name =name


# using positional attributes
child1 = Child(hair_color = "black", height = 150, eye_color ="brown", name = "puffin", skin_tone = "white")
print(f"{child1.name} hair color is {child1.hair_color}")
print(f"{child1.name} height is {child1.height}")
print(f"{child1.name} skin tone is {child1.skin_tone}")

    
```

```python
# Multilevel Inheritance = inherit from parent which inherits from another parent
# C(B) <- B(A) <- A

class Animal:
    def eat(self):
        print("This animal is eating")

    def sleep(self):
        print("This animal is sleeping")

class Prey(Animal): # -> prey inherit from animal class   
    def flee(self):
        print("This animal is fleeing")

class Predator(Animal):
    def haunt(self):
        print("This animal is haunting")

class Rabbit(Prey):
    pass

class Hawk(Predator):
    pass

class Fish(Prey, Predator):
    pass

rabbit = Rabbit()
hawk = Hawk()
fish = Fish()

rabbit.eat()
rabbit.sleep()   
rabbit.flee() 

print()
hawk.haunt()            

```
