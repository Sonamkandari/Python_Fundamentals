```python
# Interface 
# This is one of the most confusing topics because python doesn't have a separate interface keyword like java or C#. 
# Instead, python typically uses Abstract Classes (ABCs) to achieve interface-like behavior

# ->The rule is the same, but the implementation is different.
# ->An Interface is exactly like that.

# Every class must have these methods
# It does not say how those methods should work

# Real-Life Example
# Think about a mobile charger.
# Every phone supports:
# -> Charge
# -> Disconnect
# -> But internally:

# -> Samsung charges differently.
# -> Apple charges differently.
# -> OnePlus charges differently.
# The charging port follows the same contract.
# That's an interface.
```

```python
# An Interface defines a contract that specifies what methods a class must implement,
#  without specifying how those methods should be implemented

# -> Why Do we need Interfaces ?
# -> Imagine a company building payment apps
# Without interface
# Dev A -  Pay_now() , Dev B - make_payment, Dev C - pay()

# Now every class has different method names. and The project becomes difficult to maintain.

# Instead of that define interface or in python Instead, we use an Abstract Base Class.



```

```python
from abc import ABC, abstractmethod

class Payment(ABC):

    @abstractmethod
    def pay(self):
        pass

    @abstractmethod
    def refund(self):
        pass

class GooglePay(Payment):

    def pay(self):
        print("Google Pay Payment")

    def refund(self):
        print("Google Pay Refund")    

class PhonePe(Payment):

    def pay(self):
        print("PhonePe Payment")

    def refund(self):
        print("PhonePe Refund")  

# calling the constructor or created object and instance
googlePay  = GooglePay()  
googlePay.pay()    
googlePay.refund()
print()

phonePe = PhonePe()
phonePe.pay()
phonePe.refund()


```
