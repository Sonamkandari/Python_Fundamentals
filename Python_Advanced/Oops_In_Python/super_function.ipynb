```python
# super() = Function used in a child class to call method  from a parent class (superclass).
#           Allows you to extend the functionality of the inherited methods


class Shape:
    def __init__(self,color, is_filled):
        self.color = color
        self.is_filled = is_filled
        
# it will be so difficult if we have to make changes in each and every attributes even though they are common in all children classes
# there for we will declare those attributes in the parents class so that children classes can inherit them 
# it will make code more reusable and make it more efficient

class Circle(Shape):
    # we will define our constructor
    # The constructor __init__ is a special method in python known as an initializer 
    def __init__(self, color, is_filled, radius):
        # self.color = color
        # self.is_filled = is_filled  -> instead of these two line we just have to call the constructor of supper class
        super().__init__(color, is_filled)
        self.radius = radius
        self. area = 3.4 * (radius * radius)

class Rectangle(Shape):
    def __init__(self, color, is_filled, width, length):
        # self.is_filled = is_filled
        self.length = length
        super().__init__(color, is_filled)
        self.width = width
        self.area = length * width

class Square(Shape):
    def __init__( self,color, is_filled, side):
        # self.color = color
        # self.is_filled = is_filled
        self.side = side
        super().__init__(color, is_filled)
        self. area = side * side
      

class Triangle(Shape):
    def __init__(self, color, is_filled, height, width, ):  
        # self.color = color
        # self.is_filled = is_filled
        super().__init__(color, is_filled)
        self.height = height
        self.width = width
        self.area = 1//2 * width * height


# calling the constructor for the Circle and creating object and instances              
circle = Circle(color="Red",is_filled=True,radius=2)
print(circle.color)
print(circle.is_filled)
print(circle.radius)
print(circle.area)

print()
# calling the constructor for the rectangle creating object and instances
rectangle  = Rectangle(color="yellow", is_filled=False, width=9,length=10)
print (f"color of Rectangle is: {rectangle.color}")
print (f"Rectangle is is filled: {rectangle.is_filled}")
print (f"Length of Rectangle is: {rectangle.length}")
print (f"Width of Rectangle is: {rectangle.width}")
print (f"Area of Rectangle is: {rectangle.area}")



# -> State Initialization: When a new object (an instance of Circle) is created,
#  __init__ automatically runs to set up the initial state (the attributes) 
# of that specific object.

# Guaranteed Properties: Defining __init__ ensures that every Circle instance is created
#  with all the required data it needs from the moment it is instantiated.

```
