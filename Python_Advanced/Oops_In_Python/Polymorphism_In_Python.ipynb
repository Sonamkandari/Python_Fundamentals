```python
# Polymorphism = It is a Greek word that means to "one thing, many forms"
# Poly = Many 
# Morphe = form

# Two Ways to achieve polymorphism 
#  1. Inheritance = An object could be treated of the same type as a parent class
#  2. "Duck typing" = Object must have necessary attributes / methods

# simple example : Suppose your mother says:
# "Draw "  -> you draw on paper, A painter draws n canvas, An architect draws a building
# -> The wor "Draw " is the same the action changes depending on who performs

# Polymorphism is the ability of same method, or function or interface 
# exhibit different behaviors depending on the object that invokes it

from abc import ABC, abstractmethod

class Shape:
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return 3.24 * self.radius    

class Square(Shape):
    def __init__(self, width):
        self.width = width
    def area(self):
        return self.width * self.width    

class Triangle(Shape):
    def __init__(self,height, base):
        self.height = height
        self.base = base
    def area(self):
        return 1/2 * self.base * self.height    

shapes = [Circle(4), Square(5),Triangle(6,7)]
for Shape in shapes:
    print(Shape.area())





```

```python
# Method Overriding
class Animal:
    def sound(self):
        print("Animal Sound")

class Dog(Animal):
    def sound(self):
        print("Bark")

    def size(self):
        print("Big size")    

d = Dog()
d.sound()    
d.size()
```

```python
# Duck typing 



class Dog:
    def speak(self):
        print("Bark")

class Human:
    def speak(self):
        print("Hello")

def talk(obj):
    obj.speak()


talk(Dog())
talk(Human())  # is human is inherited from dog
# Python doesn't care 
# It only check Speak()

# This is called Duck typing


```
