```python
# Abstract class: A class that cannot be instantiated on its own; Meant to be subclassed
# They  can contain abstract methods, which are declare but have no implementation
#       ABSTRACT CLASSES BENEFIT
# Prevents instantiation of class itself
# Require children to use inherited abstract methods

from abc import ABC, abstractmethod


class Vehicle(ABC):

    @abstractmethod
    def go(self):   # abstract methods we declare them but we does not define them
        pass   # we define them with each of the children classes

    @abstractmethod
    def stop(self):
        pass

    @abstractmethod

    def fast(self):
        pass

# -> vehicle =  Vehicle()  # Can't instantiate abstract class Vehicle with abstract methods fast, go, stop

class Car(Vehicle):
    # now this child class should implement all the abstract methods of parent class
    def go(self):
        print("Car is going")

    def stop(self):
        print("stop the car the traffic signal")

    def fast(self) :
        print("faster the car when their is not traffic on the road")


car = Car()   
car.go()           
car.stop()           
car.fast()           

print()

class Bike(Vehicle):

    def go(self):
        print("I saw a bike ")
    def stop(self):
        print("stop two willers")
    def fast(self):
        print("faster then cycles")   

bike = Bike() # calling the constructor
bike.go()  
bike.stop()
bike.fast()
```

```python
# Another example of extract classes

from abc import ABC, abstractmethod
class  Animal(ABC):

    @abstractmethod
    def sound(self):
        pass

    @abstractmethod
    def Domestic(self):
        pass

    @abstractmethod
    def size(self):
        pass

# creating sub classes which will inherit traits from parents class   

class Dog(Animal):

    # now since Animal is a abstract class
    # now we have to implement all the  abstract methods of parent class
    # Because we know that abstract parent classes only declare the abstract methods
    # they does not implement them

    def sound(self):
        print("Dog barks so loudly")  

    def Domestic(self):
        print("Yes dogs are domestic animals")

    def size(self):
        print ("dogs are generally larger in sizes, when they grows up")  

# Calling the constructor

dog = Dog()
dog.sound()
dog.Domestic()
dog.size()
```

1. What is abc?
Python does not have built-in support for abstract classes out of the box like Java or C++. Instead, Python provides a built-in module called abc (which stands for Abstract Base Classes).

2. Why do we import ABC and @abstractmethod?
ABC (The Base Class)
What it does: It is a helper class you inherit from to make your class abstract.

Why we use it: Inheriting from ABC tells Python: "This class is incomplete/abstract. Do not allow anyone to create objects directly from it."

@abstractmethod (The Decorator)
What it does: It is a decorator placed above a method inside an abstract class.

Why we use it: It acts as a mandatory contract. It tells Python: "Every child class that inherits from this parent MUST write its own implementation for this method."
