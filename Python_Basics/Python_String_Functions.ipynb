```python
# transformations
# replace()
value = "56797,89"
print(value.replace(",",".")) # ,this will help us a particular character in a particular place

# replace(old, new) swaps part of the text with something new
name = "Mountains"
print(name.replace("n","N"))

# use case can be clean Numeric format
phone_number = "239-678-789"
print(phone_number.replace("-",""))

```
