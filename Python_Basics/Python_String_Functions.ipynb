```python
# Extraction
# Indexing and slicing

# Extract the first character
text = "sonam kandari"
print(text[0])

# Extract the last character
print(text[12])
print(text[-1])

# indexing only one character from an string

date = "2026-06-20"
# Extract the year
print(date[0:4])

# Extract the month
print(date[5:7])






```

```python
# Cleaning
#clean up whitespaces
#lstrip()
# remove extra space from the left side
city = " Dehradun"
print(city.lstrip())
#rstrip()
# remove extra space from the right side
school = "DIT University   "
print(school.rstrip())

#strip()
# remove extra spaces from both the side left and right
# It remove any characters you want from the start and end - not just spaces
text = "   hello world   "
print(text.strip())

book = "########A man called ove#####"
print(book.strip("#")) # strip will remove the special symbol also 

#whitespace cleanup
# use case- Detect Extra Spaces
# Check the length before and after stript() to find unwanted space
text ="   Engineering"
print(len(text))
print(len(text.strip()))

# Case Conversions : Use case - Clean Data for Matching
text = "python PROGRAMMING"
print(text.upper())

print(text.lower())
# clean case
#lower()
#Upper()
```
