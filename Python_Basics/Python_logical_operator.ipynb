```python
# Logical operator : used to combine multiple boolean expressions

# And operator
print (4>2 and 8>3) # And operator
print (9>3 and 2>9) # false

# or operator
print (9>3 or 2>9) 
print (4>2 or 3>9)


# checks if the system is under pressure 
cpu_usage = 70
memory_usage = 95

print(cpu_usage > 90 and memory_usage > 90) # false
print(cpu_usage > 90 or memory_usage > 90) # true

# check user credentials before login
 
email = True
password = False

print (email and password) # return false because both of credentials are not available for login

# Control Mixed condition
# And has higher priority than or


```
