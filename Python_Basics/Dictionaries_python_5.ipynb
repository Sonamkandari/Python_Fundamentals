```python
# Dictionaries are also called Hash_table in other languages
# Dictionaries are generally called key value pairs
# unordered:, the items in dictionaries are stored without any index value. 
# Unique: keys in the dictionary should be unique.
# Mutable: We  can add, modify, remove key-value pairs after the creation. 
```

```python
# Constructing a Dictionary
my_dict={'key1':'value1','key2':'value2','key3':'value3'}
print(my_dict)
```
