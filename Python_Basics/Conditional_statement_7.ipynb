```python
# conditional statement in python
# if statement in python allows us to tell the computer to perform alternative action alternative actions based on a certain set of results
# verbally, we can imagine we are telling the computer
# Hey if this case happens, perform some action
# We can then expand the idea further with elif and else statement,which allow us to tell the computer
# if, elif, else statement

a=100
b=100

if a>b:
    print("A is Greater is greater tha B")
elif b>a:
    print("B is greater than A")
else:
    print("both are Equal to Each other")

# take care of indentation and in case of some modern tools like jupyter notebook or visual studio code then already take cares of indentation we don't to correct indentation manually

```
