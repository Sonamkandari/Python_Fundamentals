```python
# Break Statement
# Break
# It stops the loop immediately It jumps out and ends the loop right away

for i in range (10):
    if i==4:
        break
    print(i)

print()
names = ['sonam', 'Annu', '','Kandari']
for name in names:
    if name == '':
        print("Empty value detected")
        break
    print(f'name = {name}')
```

```python
# Continue : 
# It skips one loop cycle without stopping the loop
# skip this one and go to next one
for i in (1,3,4):
    if(i==3): # 2 will be skipped or will not be printed
        continue
    print(i)

print()
# use continue to skip bad or empty data without stopping thw whole loop
names = ['sonam', 'Annu', '','Kandari']
for name in names:
    if name == '':
        print("Empty value detected")
        continue;
    print(f'name = {name}')
```

```python
# pass
# pass is a placeholder where nothing happened
# for now.. just keep going Do nothing

print()
names = ['sonam', 'Annu', '','Kandari']
for name in names:
    if name == '':
       pass
       name = name.replace('','unknown')
    print(f'name = {name}')
```

```python
# Break vs Continue
# Real world applications
# Loop through a list of days and print only the working days skipping the weekends
Days = ['sunday','monday','tuesday','wednesday','thursday','friday','saturday']
for day in Days:
    if day =='sunday' or day == 'saturday':
        continue
    print(day)
```
