```python
# In operator
# Checks if a value is inside another value, like a string, list, tuple, or other sequence. 

name = "sonam"
print('a'  in name)

# List 
bucket = ["sona","akku","ri"]
print("sona" in bucket)

# Validate that the domain is not on the banned list. 
#domain = "edu.in"
domain = "fake.org"
banned_list = ["spam.com,","fake.org","bot.net"]
print(domain not in banned_list)


# Identity (is) Operator
#checks if two variables refer to the same object in memory

# List of operators

x= ['a','b','c','d']
y= ['a','b','c','d']

print(x==y) # true
print(x is y) # false 


q = 10
p = 10

print (p==q)    # true
print (p is q)  # true




```

```python
# where we going use this Identity (is) Operator
# Task: Validate the email address. It must be filled in and not empty. 

email = ""
print(email!="") # false

email = "sonam@gmail"
print (email!="") # true

email = None
# None means no value at all. It is unknown, means an empty, but it is known, it is string. 
print(email!= None and email !=" ")
```
