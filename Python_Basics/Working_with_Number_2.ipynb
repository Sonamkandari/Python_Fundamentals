```python
# Basic Arithmetic
# Addition
print(9+10)
# subtraction
print(9-4)
# Multiplication
print(8*6)
# power of an component
print(7**5)
print(2**2) # square of 2 is 4
print(3**3)
# Bitwise XOR
print(4^3) 
# Bitwise & operator
print(6&8)
# Bitwise Or operator
print(7|9)
# What happens on reassignment? Will python let us write it over
a=6
print(a)
a=8
print(a) # python allow us reassignment

print(a+a)

```
