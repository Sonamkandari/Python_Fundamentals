```python
# for loop act as a iterator in python
for number in range(10):
    print("Hello") # this will print hello 10 times

for i in range(10):
    print("world") # this will print world 10 times

for j in range(3):
    print("welcome Home")

# let create an list and iterate each element of the element and check is the i th element is even or odd
EleList=[7,9,5,2,26,16,92,72,53]
for element in EleList:
    if element%2==0:
        print("this is a Even number",element)
    else:
        print("this is odd number",element)
```

```python
# while loops
# The while loop in Python is one of the most general ways to perform iteration. 
# A while statement will repeatedly execute a single statement or a group of statements as long as the condition is true. 

# example
n=20
while (n>2):
    if(n>2):
        print(n)
        n=n-1
```

```python
# Python Iterator 
# An Object that lets you go through items one by one in a sequence 
# Remembers what is done, knows what is next. 

# tuple
items = (1,2,3,4,5,"hi")
# List - List can store mix value 
#items = [1,2,3,4,5,"Hi"]
# it is like accessing each item from items set
for item in items :
    print(f"Round:{item}")

# Sequences you can Loop    
characters = "Python"
for character in characters:
    print(f"Round:{character}")

# for example you want to repeat a word a hundreds times
# so it will be so irritating if we will try to do it by our own
# so for that we will use range

for number in range (100):
    print(f"numbers: {number}")

```

```python
# For Loop real world Applications
# - We use for loops to transform data, like cleaning data before processing. 
# Total sum of all the scores
scores = [90, 67,98,12,54]
total = 0
for score in scores:
    total+= score
print(total)

# As an data engineer or data scientist or as an analyst we have to do a lot of data preparations
files =[' Report.csv', 'Data.csv', ' final.TXT']
for file in files:
    file = file.strip().lower().replace('.txt', '.csv')
    print(f"processing {file}")
```
