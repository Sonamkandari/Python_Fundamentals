```python
# Set and Booleans
# There are two other object types in Python that,
# we should quickly cover that are sets and booleans. 
# Sets are unordered collections of unique elements.
# We can construct them by using the set function.
# Let's go ahead and make a set to see how it works. 
```

```python
# we created a set
x=set()
# we added a new element in the set
x.add(1)
# we printed the information from the set
x
```

```python
mySet=set()
# lets add the elements in the set
mySet.add(2)
# print the information
mySet

# list mai [] bracket
# s={}

lst=[3,1,2]
s={3,1,2}

print(lst) #[3,2,1]
print(s) # {1,2,3}(order not guaranteed)

s.add(2)
print(s)

# Note:if i will  add 2 again and again in case of set it will be not added again
# as set is used to store unique elements only

s.add(6)
print(s)

lst.append(2) # but in case of list it store duplicate elements
print(lst)
# remove function
s.remove(2)
print(s)

```

```python
# Boolean functions
print(True)
print(False)
print(type(True))
print(type(False))

# check if it is empty value or not
print(bool("hii sonam")) # will return true because Not empty
print(bool())# return false since it is empty
```

```python
email = "sonam@gmail.com"
phone = "8989087767"
username = "sona"
# allows registration
# if any field is filled
print(any([email,phone, username])) # the output is true because at least one requirement is get filled

# is there any website which required the all the details for registration
# if any one of the details is missing it will not allow us to registered
print(all([email,phone, username])) # out put is fall since all the details are not mentioned

```

```python
# isinstance(value, type) built in function
# checks if a value belongs to a certain data TYPE

print (isinstance(2004, int)) # it will return true
print (isinstance("sonam",str))# it will also return true

# built in function to check weather an string starts with
print("Hello".endswith("o"))
# built in function to check weather an string start with
print("sonam".startswith("s"))
```
