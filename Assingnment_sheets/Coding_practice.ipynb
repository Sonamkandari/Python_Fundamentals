```python
# Fibonacci series
def fibo(N):

    if N <= 1:
        return N
    # Recursive case : add two previous numbers
    else:
        return fibo(N-1)+ fibo(N-2)
        

N =7  
for i in range (1,N+1):
    print(fibo(i))
 


```
