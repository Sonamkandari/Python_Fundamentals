```python
capacity=8
WaightList=[4,3,5,1,1,2] # we declare array like this in python
count=0
Current_sum=0

# sort the weights
WaightList.sort() # we can sort array in python directly
for weight in WaightList:
    if Current_sum+weight<=capacity:
     Current_sum+=weight
     count+=1
    else:
        break  
    
print(count) 
print("total weight is:",Current_sum)

```
