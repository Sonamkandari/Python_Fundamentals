```python
# In Python, if you want to keep the data unique, we don't use lists. We use, in that case, sets. 
# Sets: It is an unordered collection of Unique values

# check if sets or ordered or not 
my_set = {10,40,30,5}
print(my_set)  # Set is unordered

# does sets allow duplicates 
my_set =  {10,40,30,5,30}
print(my_set) # no sets does not allow duplicates

# does sets allow indexing
 # print(my_set[2]) #sets does not allow indexing

# check if set allow insertion or not   or check if it is mutable or not
# my_set[2] = 9
print(my_set)  # sets dare immutable we cannot change values in sets once declared
```

```python
# Sets special methods which cannot be used in list
# Create a set first

# How to Add Values to Set
a = {10,20,30,35,78}

# Notes- index based methods do not work with sets
# print(a.append(4)) # Append not works here

a.add(78) # this add method will works for sets
print(a)

# add() -> It inserts the items somewhere in the set, but only if it is new


# Update () -> Merges another group of values (Iterable) in to the set
```
