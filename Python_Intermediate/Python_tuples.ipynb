```python
# tuples
# what is a tuple : A tuple is a ordered collection that can't be changed after creating it

# four characteristics of tuple
my_tuple = (10,67,45)
# check whether the tuple is ordered or  not 
print(my_tuple)  # tuple is ordered

# check wether it allowed duplicates or not 
my_tuple = (10,20,30,20)
print( my_tuple) # tuple allows duplicates

#  check whether tuples are indexed or not
print(my_tuple[2])  # tuple is indexed

# check wether tuples are mutable or not
 # my_tuple[2] = 800 
print(my_tuple) # tuples are immutable

```
