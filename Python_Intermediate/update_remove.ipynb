```python
letters = ['a','b','c']
letters.append('q')# this Is how things works
letters.append('t')
print(letters)

# Insert element at Specific position 
# .insert()

movies = ['abv','ert','yuo','wer']
print(movies)

# Add 'yu' at  the start
movies.insert(1,'yu')
print(movies)

# Add a new row to the end of matrix
matrix = [
    ['a','b','c'], # Row 0
    ['d','e','f'], # Row 1
    ['g','h','f']  # Row 2
]
print(matrix)
matrix.append(['w','r','f'])
# print matrix after appending
print(matrix)

# append at a specific location
matrix.insert(2,['e','9','r'])
print(matrix)


# if we want to extend the size of a specific row in the matrix

matrix[1].append('r')
matrix[2].append('q')
print(matrix)

# use insert for the same
matrix[1].insert(1,'t')
print(matrix)


# Remove element from your list or data
# clean the matrix
pets = ['dog','cat','rat']
print(pets)
letters.clear()
print(letters)

pets.remove('cat')
print(pets)

# Pop method : remove the last method
removed =pets.pop()
print(removed) # rat is in the last so rat has been removed
print(pets) # Now only dog is there

# We can also pop the specific element

books = ['a','b','c','d']
books.pop(0)
print(books) # A will be removed

# Update Items
Names = ['Ram','Priya', 'Rani',"Coco"]
Names[0] = 'Akk' # we update the value at index 0
print(Names)


# update a matrix

matrix1 = [
    ['a','b','c'],
    ['r','u','i'],
    ['y','u','i']

]
matrix1[-1] = ['x','y','z']
print(matrix1)
```
