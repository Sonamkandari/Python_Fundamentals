```python
# Deep copy is the safest way in order to create a copy. 
# copying
import copy
matrix1 = [
    ['w','e'],
    ['r','o'],
    ['u','l']
]

matrix1_copy = matrix1.copy()
matrix1.pop()
matrix1_copy[0].append('c')
print("original",matrix1)
print("cop",matrix1_copy)



# if we want create copy which is independent to the original list 
# for that we use deepcopy
print()
schools = [
    ['t','i','i'],
    ['e','t','q'],
    ['e','r','u']
]

print (schools)
matrix1_copy = copy.deepcopy(schools)
# Now if we will remove something from the original list it will not effect our copy list
schools.pop() 
print()
# print original list after r\poping the last element
print('original:',schools)
print('Copy:   ',matrix1_copy)

# Copy.deepcopy() creates a true, independent copy for all levels. 
# copy.copy() is more general than list.copy(), not limited to lists


# Testing of  Is operator
# About to check two independent lists or they are sharing something
# Using Is Operator : Checks if two variables refer to the same object

print()
import copy
original = [
    ['a','u'],
    ['q','i']
]

#Assignment
copy1 = original
print("same object?",original is copy1,"\n")
print()
# Shallow copy
copy2 = original.copy()
print("same object?",original is copy2,"\n")
print("shared Lists", original[0] is copy2[0],"\n")

print()
# Deepcopy
copy3 = copy.deepcopy(original)
print("same object?",original is copy3)
print("shared Lists", original[0] is copy3[0],"\n")
```
