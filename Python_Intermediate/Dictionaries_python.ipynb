```python
# Methods id Dictionary
# special Methods 
user = {"id":1, "age":22, "city":"Dehradun"}

# How to Access an dictionaries
print(user["city"])
print(user["age"])
print(user["id"])
# if the key is not found, python throws a keyError
# print(user["name"])

# get() : get returns the value safely, gives None if missing
# get ()-> Missing key returns None or your default value.

# checks : In Operator : Tests if the key is inside the dictionary
print ("age" in user) # returns True
print ("name " in user) # returns False

# View objects : it give you a live of the dictionary's
# keys, values, or key value pairs
print(user.keys()) # keys returns all the keys in your dictionary

# Values() : values returns all the values in your dictionary
print(user.values())
# items() : Perfect when you need key and value together. 
# For looping, transforming data, building new dictionaries, comparing, and more 

# Looping
for u in user:
    print(u,user[u])

# more modern way to iterate through dictionaries are
print()
for key, value in user.items():
    print(key,value)

# How to change an dictionary
# Add, Remove, Update
user["name"] ='sonam' # assignment operator -> we can see in the  output a new value is added in the end of the dictionary
print(user)

user["age"] = 35
# update the value of a key
user["age"] = 36
print(user)

# there is an another way of to update multiple values together in a dictionary'
# update() : it adds new keys and updates existing ones using another dictionary
user.update({"age":20,"city":"Tokiyo","id":4 })
print(user)
```

```python
# Removing data from an dictionary
# pop() : it removes a key from the dictionary and returns its value

student = {"name":"sonam","age":22,"id":4}
print(student)
age = student.pop("age")
print("removed Item", age)

# pop() : Removes the key and returns its value, or returns your default if the key is missing
student.popitem()
print(student) 
# returns and deletes the most recent key value pair from the dictionary

# fromkeys() : builds a new dictionary where all keys get the same default value
user = dict.fromkeys(["id","name","age", "city"],None)
print (user)


```
