```python
# Python provides synthetic sugar to write small loops to generate lists, sets, tuples, dictionaries in one line. 
# These are called comprehensions and greatly increase development speed and readability. 
```

```python
# how we write it in comprehension
names=["Ravi","pooja","vijay","kiran"]
print(names)
# Comprehension refers to an instant of writing logic into multiple lines.
# You can write it into a single line. 
```

```python
#but if we want to do the same thing with using loops
new_list=[]
for name in new_list:
    new_list.append("Hello"+name)
    print(new_list)
```
