```python

# Access and read
lst = ['a','b','c','d']
print(lst)
# we can access the particular index of a list
print(lst[0],lst[3])
# use negative indexes to access items from the end of a list
print(lst[-1]) # it will help you to access last element 

# Access & Read
matrix = [
    ['a','b','c'], # Row 0
    ['d','e','f'], # Row 1
    ['g','h','i'], # Row 2
]
print(matrix)
print(matrix[2]) # using this i will get last row in the output
print (matrix[1][1])
print(matrix[-2][-2])
print(matrix[2][1])

# slicing
# How to retrieve multiple items
# Get the two items from the last
print(matrix[2][:2])
```
