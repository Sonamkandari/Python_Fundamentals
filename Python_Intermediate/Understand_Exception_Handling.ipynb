```python
# The basic terminology and syntax used to handle errors in Python is the try–except statement.
# Code that may cause an exception is placed inside the try block, and the handling of that 
# exception is implemented in the except block.

# Basic Syntax
try:
     # code that may raise an exception
     x=10/0
except ZeroDivisionError:
     # exception handling code
     print("Cannot divide by Zero")     
```

```python
try:
    file = open("data.txt")
except FileNotFoundError:
    print("File not found")
else:
    print("File opened successfully")
finally:
    file.close()
```
