```python
# one of the most important topic
# also an advance concept
# To create your own custom logic we use lambda function

# lambda is a keyword
multiple = lambda x: x*2
# Variable (multiple) : stores a lambda function which doubles a number
print(multiple(2))

print()

# Note : When a lambda has two parameters, you must pass two values when calling it
add = lambda x, y: x+y
print(add(8,9))

# subtraction using lambda function
subtract =lambda a,z:a-z
print(subtract(10, 8))

# Multiplication using lambda function
multiplication = lambda w,r,t : w*r*t
print(multiplication(8,9,10))

# check an variable using lambda 
check = lambda i:i in 'python'
print(check('h'))
print(check('p'))


print()

# here we get basic understanding how lambda works
verify = lambda name:name in "sonam kandari"
print(verify('sonam'))
print(verify('Annu'))

# Lambda + Map
# Task : Prices are stored as messy strings need cleaning floats

prices = ['$23.0',"$45.9","$45.89","$89.78"]
# we need to convert data from strings to float
```
