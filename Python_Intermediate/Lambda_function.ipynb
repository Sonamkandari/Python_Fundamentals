```python
# one of the most important topic
# also an advance concept
# To create your own custom logic we use lambda function

# lambda is a keyword
multiple = lambda x: x*2
# Variable (multiple) : stores a lambda function which doubles a number
print(multiple(2))

print()

# Note : When a lambda has two parameters, you must pass two values when calling it
add = lambda x, y: x+y
print(add(8,9))

# subtraction using lambda function
subtract =lambda a,z:a-z
print(subtract(10, 8))

# Multiplication using lambda function
multiplication = lambda w,r,t : w*r*t
print(multiplication(8,9,10))

# check an variable using lambda 
check = lambda i:i in 'python'
print(check('h'))
print(check('p'))


print()

# here we get basic understanding how lambda works
verify = lambda name:name in "sonam kandari"
print(verify('sonam'))
print(verify('Annu'))

# Lambda + Map
# Task : Prices are stored as messy strings need cleaning floats

prices = ['$23.0',"$45.9","$45.89","$89.78"]
# we need to convert data from strings to float

print(list(map(lambda p:float(p.replace('$',"")),prices)))



```

```python
# Lambda + Filter
# Remove all values lower than 100

values = [120,900,67,890,45,37,896,567,432]
# Logic for the filter

print(list(filter(lambda val : val >= 100,values)))

print()
# Remove all the children whose age is less than 16 only under 16 children can play this game
children_age = [23,22,14,15,17,19,9,21,20]
print(list(filter(lambda age : age >= 15,children_age)))

print()
# Admissions is given to the only those students whose score is higher than 80
scores = [ 80, 88, 90,45,67,78,90,96,98,23]
print(list(filter(lambda score:score >=70, scores )))
```
