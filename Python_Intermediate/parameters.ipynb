```python
# In python we have three types of variables
# Parameters
# Local variables
# Global variables

# How long does it live
# where is parameter is accessible

# Global Variable: Created outside the function can be accessed anywhere
# Local Variable : Created inside the function can be accessed only inside the function




```

```python
# Control Logic Globally: A global variable behavior without changing the function
case_rule = "lower"
def clean_name(name): # parameter
    cleaned = name.strip() # local variable
    if case_rule == "lower":
        cleaned = cleaned.lower()
    print("raw",name)
    print("cleaned data",cleaned)

clean_name(" sonam  ")
# Parameter: keeps the raw value to be reused
# Local variable: holds the processed version

# parameters and local variable can be accessed only inside the functions
```

```python
# Building full clean name
def clean_name(first_name, last_name, country):
    first = first_name.strip().lower()
    last = last_name.strip().lower()
    full_name = first + " " + last
    print(full_name, "From", country)

clean_name(" Sonam  ","KAndari","India")   
 
# Positional  Argument : Values pass to the function based on their order
# Keyword Arguments : Values pass to the function based on their Names

# Rule Positional Argument: The order of arguments must match the order of the parameters

# Another way of doing this

# Keywords arguments 
# # Safety : keyword arguments are safer since they reduce mistakes with clear names
clean_name (first_name= " sonam ", last_name= " kandari ",country= "IndIa")
```

```python
# Readability : Keywords arguments are easier to read and understand than positional 

# Overhead : Keyword arguments take more time to write and maintain

# How to choose between them.? Positional vs Keyword
# -> This Really depends on the number of parameters
# Rules says that in between 2 - 3  arguments it is okay to use positional arguments

# If there is Greater then 3 Keyword Arguments
# Mixed Arguments


# Default parameter : It is a parameter that has already a value so if you dont pass anything in python uses that value automatically




```
