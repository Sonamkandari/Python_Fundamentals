```python
class Solution {
    public int thirdMax(int[] nums) {
        int n=nums.length;
        Integer firstMaxi=null;
        Integer secondMaxi=null;
        Integer thirdMaxi=null;
        boolean foundthird=false;

       

        for(int i=0;i<n;i++){
            if (firstMaxi==null || nums[i] > firstMaxi){
                firstMaxi=nums[i];
            }
        }

        

        for(int i=0;i<n;i++){
            if(nums[i]!=firstMaxi && (secondMaxi==null || nums[i]>secondMaxi)){
                 secondMaxi=nums[i];
                 
            }
           
        }
      
          for(int i=0;i<n;i++){
            if(nums[i]!=firstMaxi && nums[i]!=secondMaxi && (thirdMaxi==null || nums[i]>thirdMaxi)){
                thirdMaxi=nums[i];
                
            }
            
             
        }
     
       
        if(thirdMaxi == null){
           return firstMaxi;
        }
            
        


       return thirdMaxi;
      

       
    }
}
```
