```python
# How to Analyse and check items in the list

# List of numbers
numbers = [1,5,2,4,3,5]
# Find highest value
print("Maximum: ",max(numbers))
# Find Minimum numbers
print("Minimum: ",min(numbers))
# find summing all values # This is also used for aggregation of the list. 
print("sum:",sum(numbers))
# len(): It finds the length of a list, (number of items). 
print("Length:",len(numbers))

# returns true if all values are true

print("All:",all(numbers))

print("All:",all([1,0,5,2]))

# Returns How many times a value appears in the list
print("count:", numbers.count(5)) # number of times 5 appear in the list

# Index function: Return the position of the first occurrence of a value
print("index of:",numbers.index(5))

# Note: .index() function returns the position of the first occurrence

# Analysis & checks
# In : it checks if a value exists in a sequence
elements = [2,4,5,6,8,9,22]
print (22 in elements) # it will return true

# comparison operator
list1 = [1,2,4]
list2 = [3,2,4]
print(list1 == list2) # It will return false
```
