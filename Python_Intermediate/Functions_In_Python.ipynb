```python
# note : Before writing new code, always check if it already exists 
# Don't Reinvent the wheel
# simple examples of functions
# why we need functions

def make_coffee():
    print("start percolator")
    print("pour some hot water from the percolator")
    print("Add sugar")
    print("Add Milk")
    print("Enjoy it")

make_coffee()
print("wake up")
print()

print("work for a while")
make_coffee()
print("Have your lunch")

# Note defining a function does not execute it,we have to call it

```

```python
# How data Flows through functions
# ->  Parameters
# -> Arguments
# -> Return

# Functions shapes
# Not all functions works on the same ways

# -> Parameters: parameters are the names used in function definition that describe what data the functions expects
# -> Arguments: Actual values passed in a function call that are assigned to parameters

# Rule : Normalize Strings
# -> Remove extra spaces and make text lowercase

def clean_name():
    name = " Sonam "
    print(name.strip().lower())

clean_name() # Hardcoded values issue
# It is not reusable because it always cleans the same value


# pass the value as a parameter to handle any input

def clean_text(text):
    print(text.strip().lower())

clean_text(" HEllow World   ")
```
